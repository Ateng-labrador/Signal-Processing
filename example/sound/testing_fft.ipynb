# Analisis Suara Gamelan

```python
import numpy as np
import scipy as sp
import fftModule as ft
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.signal import find_peaks
```

```python
samplerate, data = wavfile.read("nada_5.wav")
samplerate6, data6 = wavfile.read("nada_6.wav")
```

### Suara gamelan nada 5

```python
# banyak data
len(data)
```

```python
# durasi suara
len(data) / samplerate
```

```python
# Potong data
data1 = data[60000:90000]
data2 = data1[0:5000]
data3 = data2[0:128]
```

```python
plt.figure(figsize=(7, 3))
plt.title("Gelombang suara tangga nada 5")
plt.plot(data3)
plt.grid()
```

```python
# Sampling
N = 128
fs = 128
t = np.arange(N) / fs
freq = np.arange(N) * fs / N
```

```python
fft  = ft.fft.fft(data3)
magnitude = 2 * np.abs(fft) / N
peaks, _ = find_peaks(magnitude, height=10000)
```

```python
plt.title("Spektrum Frekuensi")
plt.plot(freq, magnitude)
plt.grid()
plt.scatter(freq[peaks], magnitude[peaks], color='red')
for p in peaks:
  plt.annotate(
      f"{freq[p]:.2f}",
      xy=(freq[p], magnitude[p]),
      xytext=(0, 0),
      textcoords="offset points",
      ha="center"
  )
```

```python
ifft = ft.ifft.ifft(fft)
plt.figure(figsize=(7, 3))
plt.title("Grafik tangga nada 5")
plt.grid()
plt.plot(ifft)
```

```python
ifft[:5]
```
